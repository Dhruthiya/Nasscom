Task 1 — Analysis Using Python Basics
Store the given sensor readings using appropriate Python data structures, then perform the following tasks.
1. Display the sensor readings

2. Identify high temperature

3. Identify low battery

4. Create a function
Create a function battery_status(charge)


```python
combined_sensor_data = [
    {"Time": 0, "Temperature": 32, "Voltage": 400, "Current": 20, "Charge": 95},
    {"Time": 1, "Temperature": 35, "Voltage": 395, "Current": 22, "Charge": 88},
    {"Time": 2, "Temperature": 38, "Voltage": 390, "Current": 25, "Charge": 80},
    {"Time": 3, "Temperature": 40, "Voltage": 385, "Current": 28, "Charge": 70},
    {"Time": 4, "Temperature": 42, "Voltage": 380, "Current": 30, "Charge": 60},
    {"Time": 5, "Temperature": 45, "Voltage": 375, "Current": 32, "Charge": 48}
]
```

```python
print("Task 1: Display all battery readings")
for reading in combined_sensor_data:
    print(f"Battery Charge: {reading['Charge']}%")
```

```python
print("\nTask 2: Identify high temperature")
for reading in combined_sensor_data:
    temperature = reading['Temperature']
    if temperature > 40:
        print(f"Time: {reading['Time']}h, Temperature: {temperature}°C - High Temperature Alert")
    else:
        print(f"Time: {reading['Time']}h, Temperature: {temperature}°C - Temperature Normal")
```

```python
print("\nTask 3: Identify low battery")
for reading in combined_sensor_data:
    charge = reading['Charge']
    if charge < 50:
        print(f"Time: {reading['Time']}h, Charge: {charge}% - Low Battery Alert")
    else:
        print(f"Time: {reading['Time']}h, Charge: {charge}% - Battery Charge Normal")
```

```python
print("\nTask 4: Create battery_status function")
def battery_status(charge):
    if charge > 70:
        return "High"
    elif 50 <= charge <= 70:
        return "Medium"
    else:
        return "Low"


print(f"Battery status for 95%: {battery_status(95)}")
print(f"Battery status for 88%: {battery_status(88)}")
print(f"Battery status for 80%: {battery_status(80)}")
print(f"Battery status for 70%: {battery_status(70)}")
print(f"Battery status for 60%: {battery_status(60)}")
print(f"Battery status for 48%: {battery_status(48)}")
```

Task 2 — Analysis Using NumPy
Convert the battery data into a NumPy array. Assume the data matrix is:
Perform the following operations:

Determine the number of rows and the number of columns.
2. Calculate the average of each parameter

3. Find the maximum and minimum temperature

```python
import numpy as np


data_matrix = np.array([
    [d['Temperature'], d['Voltage'], d['Current'], d['Charge']]
    for d in combined_sensor_data
])
```

```python
array_shape = data_matrix.shape
print(f"\n1. Shape of the array: {array_shape} (rows, columns)")


averages = np.mean(data_matrix, axis=0)
print("\n2. Average of each parameter:")
print(f"   Average Temperature (T̄): {averages[0]:.2f}°C")
print(f"   Average Voltage (V̄): {averages[1]:.2f}")
print(f"   Average Current (Ī): {averages[2]:.2f}")
print(f"   Average Charge (C̄): {averages[3]:.2f}%")
```

```python
temperatures = data_matrix[:, 0]
t_max = np.max(temperatures)
t_min = np.min(temperatures)
print(f"\n3. Maximum Temperature (T_max): {t_max}°C")
print(f"   Minimum Temperature (T_min): {t_min}°C")


voltage_column = data_matrix[:, 1]
print("\n4. Extracted Voltage Column:")
print(voltage_column)
```

```python
high_temp_readings = data_matrix[data_matrix[:, 0] > 40]
print("\n5. Rows with High Temperature (T > 40 °C):")
if high_temp_readings.size > 0:
    print(high_temp_readings)
else:
    print("No high temperature readings found.")


low_battery_readings = data_matrix[data_matrix[:, 3] < 50]
print("\n6. Rows with Low Battery (C < 50 %):")
if low_battery_readings.size > 0:
    print(low_battery_readings)
else:
    print("No low battery readings found.")
```

## Task 3 — Analysis Using Linear Algebra

```python
import numpy as np

A = np.array([
    [2, 1],
    [1, 3]
])


X = np.array([
    [40],
    [25]
])

print("Task 3: Battery Stress Index Calculation (Y = A X)")
print("---------------------------------------------------")
print("Transformation Matrix A:")
print(A)
print("\nSensor Input Vector X:")
print(X)


Y = np.dot(A, X)
print("\nResulting Vector Y (Battery Stress Index):")
print(Y)
```

### Additional Linear Algebra Tasks for Matrix A

```python

A_transpose = A.T
print("1. Transpose of A (Aᵀ):")
print(A_transpose)


A_determinant = np.linalg.det(A)
print(f"\n2. Determinant of A (det(A)): {A_determinant:.2f}")


if A_determinant != 0:
    A_inverse = np.linalg.inv(A)
    print("\n3. Inverse of A (A⁻¹):")
    print(A_inverse)
else:
    print("\n3. Inverse does not exist (determinant is zero).")
```

### Solving a System of Equations (A X = B)

```python
B = np.array([
    [100],
    [120]
])

print("System of Equations: A X = B")
print("Matrix A:")
print(A)
print("\nVector B:")
print(B)


if np.linalg.det(A) != 0:
    solution_X = np.linalg.solve(A, B)
    print("\nSolution for X (x, y) using np.linalg.solve:")
    print(solution_X)
    print(f"Therefore, x = {solution_X[0][0]:.2f} and y = {solution_X[1][0]:.2f}")
else:
    print("\nCannot solve the system of equations (matrix A is singular).")
```

## Task 4 — Analysis Using Calculus

```python
from sympy import symbols, diff

# Define the symbolic variable t
t = symbols('t')

# Define the battery charge function C(t)
C_t = 100 - 2*t - (1/2)*t**2

print("Battery Charge Function C(t):")
print(C_t)
```

```python
# 1. Find the rate of change of battery charge (dC/dt)
dC_dt = diff(C_t, t)

print("1. Rate of Change of Battery Charge (dC/dt):")
print(dC_dt)
```

```python
# 2. Calculate the discharge rate at t = 2 and t = 5
discharge_rate_t2 = dC_dt.subs(t, 2)
discharge_rate_t5 = dC_dt.subs(t, 5)

print("2. Discharge Rate Calculations:")
print(f"   dC/dt at t = 2 hours: {discharge_rate_t2}")
print(f"   dC/dt at t = 5 hours: {discharge_rate_t5}")
```

### 4.3. Interpretation of the Result

- At `t = 2` hours, the rate of change of battery charge is `-4` %/hour.
- At `t = 5` hours, the rate of change of battery charge is `-7` %/hour.

Since the absolute value of the discharge rate (`|dC/dt|`) is increasing over time (from 4 to 7), this indicates that the **battery is losing charge at an increasing rate**.
